```python
def make_paired_edits(df_in):
    multi_edit = pd.DataFrame(columns=['Gene', 'Edited_seq_num', 'Edited_Index', 'Fasta_seq'])
    
    # iterate over each gene group
    for gene, group in df_in.groupby('Gene'):
        
        # Concatenate Edited_seq_num, Edited_Index
        edited_seq_num = ','.join(map(str, group['Edited_seq_num']))
        edited_index = ','.join(map(str,group['Edited_Index']))
        
        # Bring First entry of fasts_seq for each group 
        fasta_seq = group['Fasta_seq'].iloc[0]
        
        # Bring chr, start,end,esid,strand, name, gene_frequency info for each group 
        Chr = group['Chr'].iloc[0]
        Start = group['Start'].iloc[0]
        End = group['End'].iloc[0]
        Name = group['Name'].iloc[0]
        Gene_Frequency = group['Gene_Frequency'].iloc[0]
        
        # Make new rows with the aggregated values
        multi_edit = multi_edit.append({'Gene': gene, 'Chr':Chr, 'Start':Start,'End':End,
                                        'Name':Name, 'Gene_Frequency':Gene_Frequency,
                                        'Edited_seq_num': edited_seq_num,'Edited_Index': edited_index, 
                                        'Fasta_seq': fasta_seq},
                                        ignore_index=True)
        
        # Fix Start, End, Gene_Frequency col entries 
        columns_to_convert = ['Start', 'End', 'Gene_Frequency']
        
        # Remove last ".0"
        multi_edit[columns_to_convert] = multi_edit[columns_to_convert].astype(str).replace(r'\.0$', '', regex=True)

        # Convert to numerical values
        multi_edit[columns_to_convert] = multi_edit[columns_to_convert].apply(pd.to_numeric, errors='coerce')

    
    return multi_edit

```

```python
# Adding number tag on all Gene to seperate them as invidiaual genes - stupid but trying
def make_unique_gene_names(df_in):
    df_out = df_in.copy()  # Create a copy of the input DataFrame
    
    # Group the DataFrame by the 'Gene' column
    grouped = df_out.groupby('Gene')
    
    # Iterate over each group
    for gene, group in grouped:
        # If there are duplicate gene names, add suffixes to make them unique
        if len(group) > 1:
            # Iterate over each row in the group
            for i, (_, row) in enumerate(group.iterrows(), start=1):
                # Append suffix to the gene name
                new_gene_name = f"{gene}-{i}"
                
                # Update the gene name in the DataFrame
                df_out.at[row.name, 'Gene'] = new_gene_name
    
    return df_out
```

```python
def make_multi_edits(df_in):
    
    multi_edit = df_in
    for gene, group in multi_edit.groupby('Gene'):
        # Making editing pairs by number and index 
        for seq_num, seq_index, edited_seq in zip(group['Edited_seq_num'], 
                                                  group['Edited_Index'], 
                                                  group['Edited_seq']):
            num_list = str(seq_num).split(',')
            index_list = seq_index.split(',')
            
            # Iterate over the pairs of edited sequence number and index
            for num, index in zip(num_list, index_list):
                #if num == 'PAIR':
                    #continue 
                # seq num -> integer
                num = int(num)
                # Update the corresponding Nth position in 'Edited_seq'
                edited_seq = edited_seq[:num-1] + index + edited_seq[num:]
                # Update the current gene's 'Edited_seq' to the generated updated 'edited_seq'
                multi_edit.loc[group.index, 'Edited_seq'] = edited_seq

    # rearrange Cols 
    multi_edit= multi_edit[['Gene','Chr','Start','End','Name','Gene_Frequency','Edited_seq_num',
                            'Edited_Index','Raw_seq','Edited_seq']]
    
    return multi_edit
```

```python
# Combination (order is not important) = (n!)/((n-r)!r!)

def make_combination_rows(df_in):
    
    combination_df = pd.DataFrame(columns=['Gene', 'Edited_seq_num', 'Edited_Index'])
    
    multi_comb_edit= df_in
    # Iterate over each row - which is already each gene group 
    for index, row in multi_comb_edit.iterrows():

        seq_nums = row['Edited_seq_num'].split(',')
        indices = row['Edited_Index'].split(',')

        # Generate all possible numbers of pairs which is any possible pair -1...= 0
        for num_pairs in range(1, len(seq_nums) + 1):
            # Generate combinations of pairs for the current number
            pairs = list(combinations(zip(seq_nums, indices), num_pairs))

            # Add original row ro combinfation_df
            combination_df = combination_df.append(row, ignore_index=True)

            # Add new rows for each pair combination
            for pair in pairs:
                new_row = row.copy()
                new_row['Edited_seq_num'] = ','.join([p[0] for p in pair])
                new_row['Edited_Index'] = ','.join([p[1] for p in pair])
                combination_df = combination_df.append(new_row, ignore_index=True)

    # Sort by Gene 
    combination_df = combination_df.sort_values(by='Gene').reset_index(drop=True)

    # Just get the unique combinatiosn of "Edited_seq_num"
    combination_df = combination_df.drop_duplicates(subset='Edited_seq_num', keep='first')

    # Reassign row numbers
    combination_df.reset_index(drop=True, inplace=True)

    
    # Fix Start, End, Gene_Frequency col entries 
    columns_to_convert = ['Start', 'End', 'Gene_Frequency']
    
    # Remove last ".0"
    combination_df[columns_to_convert] = combination_df[columns_to_convert].astype(str).replace(r'\.0$', '', regex=True)

    # Convert to numerical values
    combination_df[columns_to_convert] = combination_df[columns_to_convert].apply(pd.to_numeric, errors='coerce')
    
    return combination_df
    
```

```python
def make_multi_edits(df_in):
    
    multi_edit = df_in
    for gene, group in multi_edit.groupby('Gene'):
        # Making editing pairs by number and index 
        for seq_num, seq_index, edited_seq in zip(group['Edited_seq_num'], 
                                                  group['Edited_Index'], 
                                                  group['Edited_seq']):
            num_list = seq_num.split(',')
            index_list = seq_index.split(',')
            
            # Iterate over the pairs of edited sequence number and index
            for num, index in zip(num_list, index_list):
                # seq num -> integer
                num = int(num)
                # Update the corresponding Nth position in 'Edited_seq'
                edited_seq = edited_seq[:num-1] + index + edited_seq[num:]
                # Update the current gene's 'Edited_seq' to the generated updated 'edited_seq'
                multi_edit.loc[group.index, 'Edited_seq'] = edited_seq
    
    # rearrange Cols 
    multi_edit= multi_edit[['Gene','Chr','Start','End','Name','Gene_Frequency','Edited_seq_num',
                            'Edited_Index','Raw_seq','Edited_seq']]
    
    return multi_edit
```

```python
comb_df = pd.DataFrame(columns=['Gene', 'Edited_seq_num', 'Edited_Index'])

# Iterate over each row, which already each Gene group 
for index, row in ifnb_df.iterrows():
    # Split the values in 'Edited_seq_num' and 'Edited_Index'
    seq_nums = row['Edited_seq_num'].split(',')
    indices = row['Edited_Index'].split(',')
    
    # Generate permutations of pairs
    # here, the permutation pair must be however the pairs can be maken. for ex, if there are 6 pairs generated 
    # previously, the number of pair that can be generated is (where n =6 )(n!)/((n-r)!r!), where r is the selcted 
    # number to be subset, which is all the number lower than the initial n. ( this is basically formula for Combinations)
    pairs = list(permutations(zip(seq_nums, indices), 2))
    
    # Add original row to the expanded DataFrame
    comb_df = comb_df.append(row)
    
    # Add new rows for each pair permutation
    for pair in pairs:
        new_row = row.copy()
        new_row['Edited_seq_num'] = ','.join([p[0] for p in pair])
        new_row['Edited_Index'] = ','.join([p[1] for p in pair])
        comb_df = comb_df.append(new_row, ignore_index=True)

# Sort the expanded DataFrame by 'Gene' for clarity
comb_df = comb_df.sort_values(by='Gene').reset_index(drop=True)
```

```python
# Making one edits in each row of sequence 
# only the original seq = raw seq matching seqs are used as input 
# Dont ask why there are unmatches
def apply_changes(row):
    edited_index = row['Edited_Index']
    edited_seq_num = row['Edited_seq_num']
    fasta_seq_list = list(row['Fasta_seq'])
    fasta_seq_list[edited_seq_num - 1] = edited_index
    return ''.join(fasta_seq_list)

# Iterating over the group ( categorized by "Gene")
def apply_changes_grouped(group):
    group['Edited_Fasta_seq'] = group.apply(apply_changes, axis=1)
    return group
```

```python
# Collapsing single-edited seqs into one seq by treating all seq element as individual elements
# and comparing to the rest of the nth element of the seqs in the same Gene Group 

def generate_final_sequence(df):
    final_sequence = ''
    # Iterate over each position
    for i in range(len(df['Edited_Fasta_seq'].iloc[0])):
        # Get nth element for all rows (single-edited seq)
        characters = df['Edited_Fasta_seq'].str[i]
        # If nth element of all seqs in the group is the same, bring that nth element as final seq
        # bc it means that no edit was made in that position
        if characters.nunique() == 1:
            final_sequence += characters.iloc[0]
        else:
            # If there is a single different element in that nth position,
            # it means that an edit was made in that position, so take that different element as final seq
            final_sequence += next(char for char in characters if char != characters.iloc[0])
            # final seq should have all edits made in that group in the same seq
    return final_sequence

# Iterate over the same group (Gene)
def generate_final_sequence_grouped(group):
    group['Final_Fasta_seq'] = generate_final_sequence(group)
    return group
```
